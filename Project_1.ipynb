```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Ensure df is defined before proceeding
if 'df' not in globals():
    raise NameError("The dataframe 'df' is not defined. Please run the first cell (XlpQJaD4oGSI) to upload and load your dataset before running this cell.")

# Clean column names
df.columns = df.columns.str.strip()

# Convert columns
for col in ["Sales", "Profit", "Discount", "Quantity"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Convert dates and create year
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Year"] = df["Order Date"].dt.year

# Create profit margin
df["Profit Margin"] = df["Profit"] / df["Sales"]

# Confirm dataset dates
print("Date range:", df["Order Date"].min(), "to", df["Order Date"].max())
print("Number of unique orders:", df["Order ID"].nunique())
print("Number of years:", df["Year"].nunique())

# Overall company results
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
number_of_years = df["Year"].nunique()
annual_profit = total_profit / number_of_years
annual_profit_goal = annual_profit * 0.10

print("\nOVERALL COMPANY RESULTS")
print("Total sales: $", round(total_sales, 2))
print("Total profit: $", round(total_profit, 2))
print("Average annual profit: $", round(annual_profit, 2))
print("Annual profit increase needed for 10% goal: $",
      round(annual_profit_goal, 2))

# Summary function
def summarize(data, group_columns):
    result = data.groupby(group_columns).agg(
        Orders=("Order ID", "nunique"),
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    ).reset_index()

    result["Profit Margin"] = result["Profit"] / result["Sales"]
    return result

# Profit by region
print("\nPROFIT BY REGION")
region_summary = summarize(df, ["Region"])
display(region_summary.sort_values("Profit"))

# Profit by state
print("\nPROFIT BY STATE")
state_summary = summarize(df, ["State"])
display(state_summary.sort_values("Profit"))

# Texas analysis (Updated from California)
texas = df[df["State"] == "Texas"].copy()

if len(texas) == 0:
    raise ValueError("Texas was not found. Check the spelling in the State column.")

texas["Discount Group"] = np.where(
    texas["Discount"] <= 0.20,
    "Discount <= 20%",
    "Discount > 20%"
)

print("\nTEXAS OVERALL RESULTS")
print("Unique orders:", texas["Order ID"].nunique())
print("Sales: $", round(texas["Sales"].sum(), 2))
print("Profit: $", round(texas["Profit"].sum(), 2))
print("Profit margin:",
      round(texas["Profit"].sum() / texas["Sales"].sum() * 100, 2), "%")

# Texas discount comparison
print("\nTEXAS DISCOUNT COMPARISON")
discount_summary = texas.groupby("Discount Group").agg(
    Orders=("Order ID", "nunique"),
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Average_Discount=("Discount", "mean")
).reset_index()

discount_summary["Profit Margin"] = (
    discount_summary["Profit"] / discount_summary["Sales"]
)

display(discount_summary)

# Percentage of orders above 20% discount
high_discount_orders = texas[
    texas["Discount"] > 0.20
]["Order ID"].nunique()

total_texas_orders = texas["Order ID"].nunique()

print("Percentage of Texas orders above 20% discount:",
      round(high_discount_orders / total_texas_orders * 100, 2), "%")

# Texas sub-category analysis
print("\nTEXAS PROFIT BY SUB-CATEGORY")
subcategory_summary = summarize(texas, ["Sub-Category"])

# Safely map average discount using index alignment instead of raw .values
tx_sub_discount = texas.groupby("Sub-Category")["Discount"].mean()
subcategory_summary["Average Discount"] = subcategory_summary["Sub-Category"].map(tx_sub_discount)

display(subcategory_summary.sort_values("Profit"))

# Texas sub-category and segment analysis
print("\nTEXAS SUB-CATEGORY BY CUSTOMER SEGMENT")
deep_analysis = summarize(texas, ["Sub-Category", "Segment"])

tx_deep_discount = texas.groupby(["Sub-Category", "Segment"])["Discount"].mean().reset_index()
deep_analysis = deep_analysis.merge(tx_deep_discount, on=["Sub-Category", "Segment"], how="left").rename(columns={"Discount": "Average Discount"})

display(deep_analysis.sort_values("Profit"))

# Texas yearly discount analysis
print("\nTEXAS DISCOUNT RESULTS BY YEAR")
year_discount = summarize(
    texas,
    ["Year", "Discount Group"]
)

year_discount["Profit Margin"] = (
    year_discount["Profit"] / year_discount["Sales"]
)

display(year_discount.sort_values(["Year", "Discount Group"]))

# Test whether the pattern holds across years and segments
print("\nCHALLENGE TEST: YEAR AND SEGMENT")
challenge_test = summarize(
    texas,
    ["Year", "Segment", "Discount Group"]
)

challenge_test["Profit Margin"] = (
    challenge_test["Profit"] / challenge_test["Sales"]
)

display(challenge_test.sort_values(
    ["Year", "Segment", "Discount Group"]
))

# Labeled chart for Texas Sub-Category Profit
plt.figure(figsize=(10, 6))

chart_data = texas.groupby("Sub-Category")["Profit"].sum().sort_values()

chart_data.plot(kind="barh")
plt.title("Texas Profit by Product Sub-Category")
plt.xlabel("Total Profit ($)")
plt.ylabel("Product Sub-Category")
plt.axvline(0, color="black", linewidth=0.8)
plt.tight_layout()
plt.show()

# Expected-outcome baseline
print("\nEXPECTED OUTCOME SCENARIOS")

texas_high_discount = texas[
    texas["Discount"] > 0.20
]

annual_affected_sales = (
    texas_high_discount["Sales"].sum() / number_of_years
)

annual_affected_profit = (
    texas_high_discount["Profit"].sum() / number_of_years
)

print("Annual sales affected: $", round(annual_affected_sales, 2))
print("Annual profit from affected orders: $",
      round(annual_affected_profit, 2))
print("Annual company profit-improvement goal: $",
      round(annual_profit_goal, 2))

# Scenario assumptions
scenario_results = []

for assumption in [0.05, 0.10, 0.15]:
    estimated_improvement = abs(annual_affected_profit) * assumption
    percent_of_goal = estimated_improvement / annual_profit_goal * 100

    scenario_results.append({
        "Assumed improvement": f"{int(assumption * 100)}%",
        "Estimated annual improvement": round(estimated_improvement, 2),
        "Percent of annual goal": round(percent_of_goal, 1)
    })

scenario_table = pd.DataFrame(scenario_results)
display(scenario_table)

# Automatic recommendation statement
print("\nRECOMMENDATION")

print(
    "Office Solutions should test a 20% discount cap in Texas, "
    "especially for the Texas sub-categories with the largest losses. "
    "The company should monitor profit, sales, order volume, and profit margin "
    "before expanding the policy."
)

print(
    "\nLIMITATION: The analysis shows an association between discounts and profit. "
    "It does not prove that discounts alone caused the losses. "
    "The scenario results are estimates, not causal predictions."
)
```

```python
# STEP 7: Reinforce analysis
# Focus: Texas Binders product, year, customer segment, and discount patterns.
# This cell analyzes historical performance; Step 11 estimates potential recovery.

import pandas as pd
import matplotlib.pyplot as plt

# Filter the dataset to Texas Binders
binders = df[
    (df["State"] == "Texas") &
    (df["Sub-Category"] == "Binders")
].copy()

# Extract year from Order Date if a Year column is not already available
if "Year" not in binders.columns:
    binders["Year"] = pd.to_datetime(binders["Order Date"]).dt.year

# Create discount bands for comparison
binders["Discount Band"] = binders["Discount"].apply(
    lambda discount: "20% or less" if discount <= 0.20 else "Over 20%"
)

# --------------------------------------------------
# TABLE 7.1: TEXAS BINDERS PRODUCT PROFITABILITY
# --------------------------------------------------

product_analysis = (
    binders.groupby(["Product Name"], as_index=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Average_Discount=("Discount", "mean"),
        Unique_Orders=("Order ID", "nunique")
    )
)

product_analysis["Profit Margin (%)"] = (
    product_analysis["Profit"] / product_analysis["Sales"] * 100
)

product_analysis = product_analysis.sort_values("Profit")

print("TABLE 7.1: TEXAS BINDERS PRODUCT PROFITABILITY")
display(product_analysis.round({
    "Sales": 2,
    "Profit": 2,
    "Average_Discount": 3,
    "Profit Margin (%)": 2
}))

# Chart 7.1: Products with the largest losses
largest_losses = product_analysis.head(10).sort_values("Profit")

plt.figure(figsize=(11, 6))
plt.barh(largest_losses["Product Name"], largest_losses["Profit"])
plt.axvline(0, linewidth=1)
plt.title("Figure 7.1: Texas Binders Products with the Largest Profit Losses")
plt.xlabel("Total Profit ($)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()

# --------------------------------------------------
# TABLE 7.2: YEAR x CUSTOMER SEGMENT x DISCOUNT BAND
# --------------------------------------------------

cross_analysis = (
    binders.groupby(
        ["Year", "Segment", "Discount Band"],
        as_index=False
    )
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Unique_Orders=("Order ID", "nunique"),
        Average_Discount=("Discount", "mean")
    )
)

cross_analysis["Profit Margin (%)"] = (
    cross_analysis["Profit"] / cross_analysis["Sales"] * 100
)

print("TABLE 7.2: TEXAS BINDERS BY YEAR, CUSTOMER SEGMENT, AND DISCOUNT BAND")
display(cross_analysis.sort_values(
    ["Year", "Segment", "Discount Band"]
).round({
    "Sales": 2,
    "Profit": 2,
    "Average_Discount": 3,
    "Profit Margin (%)": 2
}))

# Chart 7.2: Profit by year, segment, and discount band
chart_data = cross_analysis.copy()
chart_data["Group"] = (
    chart_data["Year"].astype(str) + " | " +
    chart_data["Segment"] + " | " +
    chart_data["Discount Band"]
)

plt.figure(figsize=(13, 7))
plt.bar(chart_data["Group"], chart_data["Profit"])
plt.axhline(0, linewidth=1)
plt.title("Figure 7.2: Texas Binders Profit by Year, Segment, and Discount Band")
plt.xlabel("Year | Customer Segment | Discount Band")
plt.ylabel("Total Profit ($)")
plt.xticks(rotation=75, ha="right")
plt.tight_layout()
plt.show()

# --------------------------------------------------
# STEP 7: SUMMARY CHECKS
# --------------------------------------------------

print("STEP 7 SUMMARY")
print(f"Texas Binders sales: ${binders['Sales'].sum():,.2f}")
print(f"Texas Binders profit: ${binders['Profit'].sum():,.2f}")
print(
    "Texas Binders profit margin: "
    f"{binders['Profit'].sum() / binders['Sales'].sum() * 100:.2f}%"
)
print(f"Average discount: {binders['Discount'].mean():.1%}")
print(f"Unique orders: {binders['Order ID'].nunique():,}")
```

This codeblock helps us investigate why Texas Binders are losing money and whether the losses are unique to products, year, customer groups, or discount amount. This codeblock filters the data, calculates profitability, visualizes the largest losses via bar chart, and compares profit across 2017-2020, customer segments, and discount amounts. In short: total sales, profit, profit margin, average discount, and unique orders for Texas Binders.

```python
# Step 1: Filter for orders affected by the recommendation
affected = df[
    (df["State"] == "Texas") &
    (df["Sub-Category"] == "Binders") &
    (df["Discount"] > 0.20)
]

# Step 2: Calculate historical totals
total_sales = affected["Sales"].sum()
total_profit = affected["Profit"].sum()

# Dataset covers 2017–2020 (4 years)
years = 4
annual_sales = total_sales / years
annual_profit = total_profit / years

print("=== Texas Binders: Orders with Discounts Above 20% ===")
print(f"Total sales (2017–2020): ${total_sales:,.2f}")
print(f"Total profit (2017–2020): ${total_profit:,.2f}")
print(f"Average annual sales: ${annual_sales:,.2f}")
print(f"Average annual profit: ${annual_profit:,.2f}")

# Step 3: Estimate potential improvement
# These recovery rates are illustrative assumptions.
recovery_rates = [0.05, 0.10, 0.20]

company_target = 7159.93

# Only estimate loss recovery if the affected group has a historical loss.
if annual_profit < 0:
    annual_loss = abs(annual_profit)

    print("\n=== Scenario Estimates ===")

    for rate in recovery_rates:
        improvement = annual_loss * rate
        target_percentage = (improvement / company_target) * 100

        print(f"\nAssumed recovery: {rate:.0%}")
        print(f"Estimated annual improvement: ${improvement:,.2f}")
        print(f"Percentage of company target: {target_percentage:.2f}%")

else:
    print(
        "\nThe affected group has positive or zero historical profit. "
        "Do not calculate loss recovery from this group. "
        "Review the results and use a different scenario method."
    )
```
